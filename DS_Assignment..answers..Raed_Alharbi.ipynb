## List Comprehensions (2 points)
1. Create a list called `cubes` containing the cubes of the numbers 1 to 30 (inclusive).
2. From `cubes`, create a new list called `cubesOf3` containing only the cubes of numbers that are multiples of 3.
3. Using a **single list comprehension**, create a list called `fizzbuzz` for numbers 1 to 30 where:

   - multiples of 3 (only) are replaced by `'Fizz'`

````python
cubes = [x**3 for x in range(1, 31)]
print(cubes)
cubesOf3 = [c for c in cubes if c % 3 == 0]
print(cubesOf3)
fizzbuzz = ['FizzBuzz' if x % 15 == 0 else 'Fizz' if x % 3 == 0 else 'Buzz' if x % 5 == 0 else x for x in range(1, 31)]
print(fizzbuzz)
````

## Zip (2 points)
Given the two lists below:
```python
students = ['Ahmad', 'Sara', 'Faisal', 'Noura', 'Omar']
scores   = [88, 95, 72, 91, 67]
```
1. Use `zip` to create a list of tuples called `studentScores`, e.g. `[('Ahmad', 88), ...]`.
2. Iterate over `studentScores` using `enumerate` and print each entry as: `1. Ahmad scored 88`.
3. Without using the built-in `max()` on `scores` directly, iterate over `studentScores` to find and print the name and score of the student with the **highest** score.

````python
students = ['Ahmad', 'Sara', 'Faisal', 'Noura', 'Omar']
scores   = [88, 95, 72, 91, 67]

studentScores = list(zip(students, scores))
print(studentScores)
print()

for i, (name, score) in enumerate(studentScores, start=1):
    print(f"{i}. {name} scored {score}")

highest_score = studentScores[0][1]
top_student   = studentScores[0][0]

for name, score in studentScores:
    if score > highest_score:
        highest_score = score
        top_student   = name

print(f"Top student: {top_student} with a score of {highest_score}")
````

**The output generated by the code written to answer this question should be as follows:**

[('Ahmad', 88), ('Sara', 95), ('Faisal', 72), ('Noura', 91), ('Omar', 67)]
1. Ahmad scored 88
2. Sara scored 95
3. Faisal scored 72
4. Noura scored 91
5. Omar scored 67
Top student: Sara with a score of 95

## Dictionaries (3 points)
Given the two lists below:
```python
products = ['Laptop', 'Mouse', 'Keyboard', 'Monitor', 'Headset']
prices   = [3200, 45, 90, 750, 120]
```
1. Create a dictionary called `priceBook` mapping each product to its price.
2. Loop over `priceBook` and print each item as: `The price of Laptop is 3200 SAR.`
3. Compute and print the **total** price of all products using `priceBook`.
4. Add a new product `'Webcam'` priced at `150` to `priceBook`.
5. Remove `'Mouse'` from `priceBook`.

````python
products = ['Laptop', 'Mouse', 'Keyboard', 'Monitor', 'Headset']
prices   = [3200, 45, 90, 750, 120]

priceBook = dict(zip(products, prices))
print(priceBook)
print()

for k, v in priceBook.items():
    print(f"The price of {k} is {v} SAR.")
print()

total_price = sum(priceBook.values())
print(f"Total price of all products: {total_price} SAR")
print()

priceBook['Webcam'] = 150
print(priceBook)

priceBook.pop('Mouse')
print(priceBook)
````

## String Methods (3 points)
Given the text below:
```python
text = """python is a general purpose language. it is widely used in data science.
many companies rely on python for automation and analysis."""
```
1. Split the text into a list of sentences (hint: split on `'.'` and remove the empty trailing element).
2. Create a new list where each sentence is stripped of extra spaces, capitalized, and has `'python'` replaced with `'Python'`.
3. Join the cleaned sentences back into a single text (each sentence on its own line) and print it.
4. Count how many times the word `'python'` (case-insensitive) appears in the **original** text.

````python
text = """python is a general purpose language. it is widely used in data science.
many companies rely on python for automation and analysis."""

sentences = text.split('.')
sentences.pop(-1)
print(sentences)

cleaned_sentences = [s.strip().capitalize().replace('python', 'Python') + '.' for s in sentences]
print(cleaned_sentences)

newText = "\n".join(cleaned_sentences)
print(newText)

python_count = text.lower().count('python')
print(f"'python' appears {python_count} times.")
````

## Lambda & Map (2 points)
Given the list of Celsius temperatures below:
```python
celsius = [0, 15, 20, 30, 37, 100]
```
1. Using `lambda` and `map`, create a list `fahrenheit` converting each value using the formula `F = C * 9/5 + 32`.
2. Using `lambda` and `map`, create a list `labels` where each Fahrenheit value is replaced with the string `'Hot'` if it is greater than or equal to 90, and `'Not Hot'` otherwise.

````python
celsius = [0, 15, 20, 30, 37, 100]

fahrenheit = list(map(lambda c: c * 9/5 + 32, celsius))
print(fahrenheit)

labels = list(map(lambda f: 'Hot' if f >= 90 else 'Not Hot', fahrenheit))
print(labels)
````

['Not Hot', 'Not Hot', 'Not Hot', 'Not Hot', 'Hot', 'Hot']

## NumPy — Sorting, Broadcasting & Masking (3 points)
Using the same array `X` from B7:
1. Sort each **row** of `X` independently (ascending) without modifying the sorting axis of the columns.
2. Create a 1D array `add_row = np.array([1, 0, -1, 2])` and use broadcasting to add it to every row of `X`.
3. Using masking, extract all elements of `X` that are strictly greater than 10.
4. Compute the sum of all elements in `X` that are even, using masking (no loops).

````python
import numpy as np

np.random.seed(42)
x = np.random.randint(0, 20, size=12)
X = x.reshape(3, 4)
print("X:", X)

row_sorted_X = np.sort(X, axis=1)
print("Row-sorted X:", row_sorted_X)

add_row = np.array([1, 0, -1, 2])
broadcast_res = X + add_row
print("X + add_row (broadcast):", broadcast_res)

greater_than_10 = X[X > 10]
print("Elements > 10:", greater_than_10)

sum_even = X[X % 2 == 0].sum()
print("Sum of even elements:", sum_even)

````
